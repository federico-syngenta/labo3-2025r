```python
# =============================================================================
# PARTE 2: FUNCIONES UTILITARIAS
# =============================================================================

def display_dataframe_info(df, name="DataFrame"):
    """
    Muestra información detallada sobre un DataFrame
    """
    print(f"\n{'='*50}")
    print(f"📊 INFORMACIÓN DE {name.upper()}")
    print(f"{'='*50}")
    print(f"🔹 Forma: {df.shape}")
    print(f"🔹 Memoria utilizada: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print(f"🔹 Rango de fechas: {df['periodo'].min()} a {df['periodo'].max()}" if 'periodo' in df.columns else "")
    print(f"\n📋 Tipos de datos:")
    print(df.dtypes.value_counts())
    print(f"\n🔍 Valores nulos:")
    null_counts = df.isnull().sum()
    if null_counts.sum() > 0:
        print(null_counts[null_counts > 0])
    else:
        print("✅ No hay valores nulos")
    
    if 'tn' in df.columns:
        print(f"\n📈 Estadísticas de 'tn':")
        print(f"   • Total: {df['tn'].sum():.2f}")
        print(f"   • Promedio: {df['tn'].mean():.4f}")
        print(f"   • Mediana: {df['tn'].median():.4f}")
        print(f"   • Ceros: {(df['tn'] == 0).sum():,} ({(df['tn'] == 0).mean()*100:.1f}%)")

def validate_date_format(df, date_col='periodo'):
    """
    Valida y convierte formato de fecha
    """
    print(f"🕐 Validando formato de fecha en columna '{date_col}'...")
    
    if df[date_col].dtype == 'object':
        # Intentar convertir formato YYYYMM a datetime
        try:
            df[date_col] = pd.to_datetime(df[date_col], format='%Y%m')
            print("✅ Fecha convertida desde formato YYYYMM")
        except:
            try:
                df[date_col] = pd.to_datetime(df[date_col])
                print("✅ Fecha convertida con auto-detección")
            except:
                print("❌ Error al convertir fechas")
                return False
    
    print(f"📅 Rango de fechas: {df[date_col].min()} a {df[date_col].max()}")
    return True

def get_product_customer_lifecycles(sales_df):
    """
    Calcula los ciclos de vida de productos y clientes
    """
    print("🔍 Calculando ciclos de vida de productos y clientes...")
    
    # Ciclo de vida de productos
    product_lifecycle = sales_df.groupby('product_id')['periodo'].agg(['min', 'max']).reset_index()
    product_lifecycle.columns = ['product_id', 'product_first_sale', 'product_last_sale']
    
    # Ciclo de vida de clientes
    customer_lifecycle = sales_df.groupby('customer_id')['periodo'].agg(['min', 'max']).reset_index()
    customer_lifecycle.columns = ['customer_id', 'customer_first_purchase', 'customer_last_purchase']
    
    # Ciclo de vida de combinaciones producto-cliente
    product_customer_lifecycle = sales_df.groupby(['product_id', 'customer_id'])['periodo'].agg(['min', 'max']).reset_index()
    product_customer_lifecycle.columns = ['product_id', 'customer_id', 'first_purchase', 'last_purchase']
    
    print(f"✅ Productos únicos: {len(product_lifecycle):,}")
    print(f"✅ Clientes únicos: {len(customer_lifecycle):,}")
    print(f"✅ Combinaciones producto-cliente: {len(product_customer_lifecycle):,}")
    
    return product_lifecycle, customer_lifecycle, product_customer_lifecycle

def calculate_metrics(y_true, y_pred, model_name="Modelo"):
    """
    Calcula métricas de evaluación
    """
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    
    # Métrica de error porcentual personalizada
    error_percentage = np.abs(y_pred - y_true).sum() / y_true.sum() * 100
    
    metrics = {
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2': r2,
        'Error_Percentage': error_percentage
    }
    
    print(f"\n📊 MÉTRICAS DE {model_name.upper()}")
    print("-" * 40)
    for metric, value in metrics.items():
        print(f"{metric}: {value:.4f}")
    
    return metrics

def plot_predictions_vs_actual(y_true, y_pred, model_name="Modelo", sample_size=1000):
    """
    Grafica predicciones vs valores reales
    """
    # Tomar muestra si hay muchos datos
    if len(y_true) > sample_size:
        indices = np.random.choice(len(y_true), sample_size, replace=False)
        y_true_sample = y_true.iloc[indices] if hasattr(y_true, 'iloc') else y_true[indices]
        y_pred_sample = y_pred[indices]
    else:
        y_true_sample = y_true
        y_pred_sample = y_pred
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Scatter plot
    ax1.scatter(y_true_sample, y_pred_sample, alpha=0.6, s=30)
    max_val = max(y_true_sample.max(), y_pred_sample.max())
    ax1.plot([0, max_val], [0, max_val], 'r--', linewidth=2)
    ax1.set_xlabel('Valores Reales')
    ax1.set_ylabel('Valores Predichos')
    ax1.set_title(f'{model_name} - Predicho vs Real')
    ax1.grid(True, alpha=0.3)
    
    # Histograma de errores
    errors = y_pred_sample - y_true_sample
    ax2.hist(errors, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
    ax2.axvline(x=0, color='red', linestyle='--', linewidth=2)
    ax2.set_xlabel('Error de Predicción')
    ax2.set_ylabel('Frecuencia')
    ax2.set_title(f'{model_name} - Distribución de Errores')
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

print("✅ Funciones utilitarias definidas correctamente")
```

```python
# =============================================================================
# PARTE 3: LECTURA Y CARGA DE DATASETS
# =============================================================================

def load_datasets(data_path="../datasets/"):
    """
    Carga todos los datasets necesarios
    """
    print("📂 Iniciando carga de datasets...")
    
    try:
        # Cargar datos de ventas (sell-in)
        print("🔄 Cargando datos de ventas (sell-in.txt)...")
        sales = pd.read_csv(f"{data_path}sell-in.txt", 
                           sep="\t", 
                           dtype={"periodo": str})
        
        # Cargar datos de stock
        print("🔄 Cargando datos de stock (tb_stocks.txt)...")
        stocks = pd.read_csv(f"{data_path}tb_stocks.txt", 
                            sep="\t", 
                            dtype={"periodo": str})
        
        # Cargar información de productos
        print("🔄 Cargando información de productos (tb_productos.txt)...")
        product_info = pd.read_csv(f"{data_path}tb_productos.txt", sep="\t")
        
        # Cargar productos a predecir
        print("🔄 Cargando productos a predecir (product_id_apredecir201912.txt)...")
        products_to_predict = pd.read_csv(f'{data_path}product_id_apredecir201912.txt')
        
        print("✅ Todos los datasets cargados correctamente")
        
        return sales, stocks, product_info, products_to_predict
        
    except FileNotFoundError as e:
        print(f"❌ Error: No se pudo encontrar el archivo {e}")
        return None, None, None, None
    except Exception as e:
        print(f"❌ Error al cargar datasets: {e}")
        return None, None, None, None

def preprocess_datasets(sales, stocks, product_info, products_to_predict):
    """
    Preprocesa los datasets básicos
    """
    print("\n🔧 Iniciando preprocesamiento de datasets...")
    
    # Convertir fechas
    print("📅 Convirtiendo formatos de fecha...")
    validate_date_format(sales, 'periodo')
    validate_date_format(stocks, 'periodo')
    
    # Mostrar información inicial
    display_dataframe_info(sales, "SALES (Original)")
    display_dataframe_info(stocks, "STOCKS")
    display_dataframe_info(product_info, "PRODUCT_INFO")
    
    # Verificar productos a predecir
    print(f"\n🎯 Productos a predecir: {len(products_to_predict):,}")
    if 'product_id' in products_to_predict.columns:
        predict_products_list = products_to_predict['product_id'].unique()
    else:
        # Si la columna tiene otro nombre, intentar la primera columna
        predict_products_list = products_to_predict.iloc[:, 0].unique()
    
    print(f"📋 Lista de productos a predecir (primeros 10): {predict_products_list[:10]}")
    
    # Validar que los productos a predecir estén en los datos históricos
    products_in_sales = sales['product_id'].unique()
    missing_products = set(predict_products_list) - set(products_in_sales)
    
    if missing_products:
        print(f"⚠️  Productos a predecir que NO están en ventas históricas: {len(missing_products)}")
        print(f"    Ejemplos: {list(missing_products)[:5]}")
    else:
        print("✅ Todos los productos a predecir tienen datos históricos")
    
    # Agregar información de productos a ventas
    print("🔗 Fusionando información de productos con ventas...")
    sales_with_info = sales.merge(product_info, on='product_id', how='left')
    
    # Verificar merge
    sales_without_info = sales_with_info[sales_with_info['cat1'].isnull()]
    if len(sales_without_info) > 0:
        print(f"⚠️  Ventas sin información de producto: {len(sales_without_info):,}")
        print(f"    Productos únicos sin info: {sales_without_info['product_id'].nunique()}")
    else:
        print("✅ Todas las ventas tienen información de producto")
    
    # Agregar stocks a ventas
    print("🔗 Fusionando información de stocks con ventas...")
    sales_complete = sales_with_info.merge(stocks, on=['periodo', 'product_id'], how='left')
    
    # Mostrar resultado final
    display_dataframe_info(sales_complete, "SALES (Completo)")
    
    return sales_complete, predict_products_list

# Ejecutar carga de datos
print("🚀 INICIANDO CARGA Y PREPROCESAMIENTO DE DATOS")
print("=" * 60)

# Cargar datasets
sales, stocks, product_info, products_to_predict = load_datasets()

if sales is not None:
    # Preprocesar datasets
    sales_complete, predict_products_list = preprocess_datasets(sales, stocks, product_info, products_to_predict)
    
    print(f"\n🎉 CARGA COMPLETADA EXITOSAMENTE")
    print(f"📊 Dataset final: {sales_complete.shape[0]:,} registros")
    print(f"🎯 Productos a predecir: {len(predict_products_list):,}")
    print(f"📅 Rango temporal: {sales_complete['periodo'].min()} a {sales_complete['periodo'].max()}")
else:
    print("❌ Error en la carga de datos. Verificar paths y archivos.")
```

# Dealing with zeros

```python
# =============================================================================
# PARTE 4: AGREGADO DE CEROS (LÓGICA COMPLEJA)
# =============================================================================

def create_zero_transactions(sales_df, verbose=True):
    """
    Crea transacciones con tn=0 siguiendo la lógica compleja de ciclos de vida
    
    Lógica:
    - Un producto existe desde su primera venta hasta su última venta
    - Un cliente existe desde su primera compra hasta su última compra  
    - Para cada combinación <product_id, customer_id> que esté "viva" en un período,
      si no existe transacción, se inserta una con tn=0
    """
    print("🔧 INICIANDO PROCESO DE AGREGADO DE CEROS")
    print("=" * 60)
    
    # Paso 1: Calcular ciclos de vida
    if verbose:
        print("📊 Calculando ciclos de vida...")
    
    # Ciclo de vida de productos (primera a última venta)
    product_lifecycle = sales_df.groupby('product_id')['periodo'].agg(['min', 'max']).reset_index()
    product_lifecycle.columns = ['product_id', 'product_start', 'product_end']
    
    # Ciclo de vida de clientes (primera a última compra)
    customer_lifecycle = sales_df.groupby('customer_id')['periodo'].agg(['min', 'max']).reset_index()
    customer_lifecycle.columns = ['customer_id', 'customer_start', 'customer_end']
    
    if verbose:
        print(f"   • Productos únicos: {len(product_lifecycle):,}")
        print(f"   • Clientes únicos: {len(customer_lifecycle):,}")
        print(f"   • Rango temporal: {sales_df['periodo'].min()} a {sales_df['periodo'].max()}")
    
    # Paso 2: Crear grid completo de períodos
    all_periods = pd.date_range(
        start=sales_df['periodo'].min(),
        end=sales_df['periodo'].max(),
        freq='MS'  # Month Start frequency
    )
    
    if verbose:
        print(f"   • Períodos totales: {len(all_periods)}")
    
    # Paso 3: Determinar combinaciones válidas por período
    print("🔄 Generando combinaciones válidas producto-cliente por período...")
    
    valid_combinations = []
    
    for period in tqdm(all_periods, desc="Procesando períodos"):
        # Productos vivos en este período
        active_products = product_lifecycle[
            (product_lifecycle['product_start'] <= period) & 
            (product_lifecycle['product_end'] >= period)
        ]['product_id'].values
        
        # Clientes vivos en este período  
        active_customers = customer_lifecycle[
            (customer_lifecycle['customer_start'] <= period) & 
            (customer_lifecycle['customer_end'] >= period)
        ]['customer_id'].values
        
        # Crear todas las combinaciones válidas para este período
        if len(active_products) > 0 and len(active_customers) > 0:
            period_combinations = pd.DataFrame({
                'periodo': period,
                'product_id': np.repeat(active_products, len(active_customers)),
                'customer_id': np.tile(active_customers, len(active_products))
            })
            valid_combinations.append(period_combinations)
    
    # Combinar todas las combinaciones
    print("🔗 Combinando todas las combinaciones válidas...")
    all_valid_combinations = pd.concat(valid_combinations, ignore_index=True)
    
    if verbose:
        print(f"   • Combinaciones teóricas totales: {len(all_valid_combinations):,}")
    
    # Paso 4: Identificar transacciones existentes
    print("🔍 Identificando transacciones existentes...")
    existing_transactions = sales_df[['periodo', 'product_id', 'customer_id']].copy()
    existing_transactions['exists'] = True
    
    # Paso 5: Encontrar combinaciones faltantes (que necesitan tn=0)
    print("❌ Identificando transacciones faltantes...")
    complete_grid = all_valid_combinations.merge(
        existing_transactions,
        on=['periodo', 'product_id', 'customer_id'],
        how='left'
    )
    
    missing_transactions = complete_grid[complete_grid['exists'].isnull()].copy()
    missing_transactions = missing_transactions.drop('exists', axis=1)
    
    if verbose:
        print(f"   • Transacciones existentes: {len(existing_transactions):,}")
        print(f"   • Transacciones faltantes: {len(missing_transactions):,}")
        print(f"   • Factor de multiplicación: {len(complete_grid) / len(existing_transactions):.2f}x")
    
    # Paso 6: Crear transacciones con tn=0
    print("➕ Creando transacciones con tn=0...")
    
    # Agregar columnas faltantes con valores por defecto
    missing_transactions['tn'] = 0.0
    missing_transactions['plan_precios_cuidados'] = 0
    missing_transactions['cust_request_qty'] = 0
    missing_transactions['cust_request_tn'] = 0.0
    
    # Agregar información de productos (merge con product_info)
    print("🔗 Agregando información de productos a transacciones faltantes...")
    
    # Obtener product_info del dataset original
    product_cols = ['product_id', 'cat1', 'cat2', 'cat3', 'brand', 'sku_size', 'descripcion']
    product_info_unique = sales_df[product_cols].drop_duplicates()
    
    missing_transactions = missing_transactions.merge(
        product_info_unique,
        on='product_id',
        how='left'
    )
    
    # Agregar stock_final como NaN (se manejará después)
    missing_transactions['stock_final'] = np.nan
    
    # Paso 7: Combinar transacciones originales con las nuevas
    print("🔄 Combinando transacciones originales con las nuevas...")
    
    # Asegurar que las columnas estén en el mismo orden
    columns_order = sales_df.columns.tolist()
    missing_transactions = missing_transactions[columns_order]
    
    # Combinar datasets
    complete_sales = pd.concat([sales_df, missing_transactions], ignore_index=True)
    
    # Ordenar por período, producto y cliente
    complete_sales = complete_sales.sort_values(['periodo', 'product_id', 'customer_id'])
    
    if verbose:
        print("\n📊 RESULTADO FINAL:")
        print(f"   • Registros originales: {len(sales_df):,}")
        print(f"   • Registros agregados (tn=0): {len(missing_transactions):,}")
        print(f"   • Total final: {len(complete_sales):,}")
        print(f"   • Factor de crecimiento: {len(complete_sales) / len(sales_df):.2f}x")
        print(f"   • Porcentaje de ceros: {(complete_sales['tn'] == 0).mean() * 100:.1f}%")
    
    return complete_sales, missing_transactions

def validate_zero_insertion(complete_sales, original_sales, verbose=True):
    """
    Valida que el proceso de inserción de ceros sea correcto
    """
    print("\n🔍 VALIDANDO INSERCIÓN DE CEROS")
    print("-" * 40)
    
    # Verificar que no se perdieron transacciones originales
    original_sum = original_sales['tn'].sum()
    new_sum = complete_sales[complete_sales['tn'] > 0]['tn'].sum()
    
    if abs(original_sum - new_sum) < 0.01:  # Pequeña tolerancia por precisión float
        print("✅ Suma de ventas originales preservada")
    else:
        print(f"❌ ERROR: Suma original: {original_sum:.2f}, Nueva suma: {new_sum:.2f}")
    
    # Verificar estructura temporal
    periods_original = set(original_sales['periodo'].unique())
    periods_new = set(complete_sales['periodo'].unique())
    
    if periods_original == periods_new:
        print("✅ Períodos temporales preservados")
    else:
        print(f"❌ ERROR: Períodos diferentes")
    
    # Estadísticas de ceros
    zero_count = (complete_sales['tn'] == 0).sum()
    total_count = len(complete_sales)
    zero_percentage = (zero_count / total_count) * 100
    
    print(f"📊 Registros con tn=0: {zero_count:,} ({zero_percentage:.1f}%)")
    print(f"📊 Registros con tn>0: {total_count - zero_count:,} ({100-zero_percentage:.1f}%)")
    
    if verbose:
        display_dataframe_info(complete_sales, "SALES (Con Ceros)")
    
    return True

# Ejecutar proceso de agregado de ceros
print("🚀 EJECUTANDO AGREGADO DE CEROS")
print("⚠️  ADVERTENCIA: Este proceso puede tomar varios minutos...")
print("📊 Dataset inicial:", sales_complete.shape)

# Ejecutar el proceso
complete_sales_with_zeros, added_zero_transactions = create_zero_transactions(
    sales_complete, 
    verbose=True
)

# Validar el resultado
validate_zero_insertion(complete_sales_with_zeros, sales_complete, verbose=True)

print("\n🎉 PROCESO DE AGREGADO DE CEROS COMPLETADO")
print(f"📈 Crecimiento: {sales_complete.shape[0]:,} → {complete_sales_with_zeros.shape[0]:,}")
```

```python
# =============================================================================
# PARTE 5: FEATURE ENGINEERING OPTIMIZADO
# =============================================================================

import gc
import psutil
import os

def print_memory_usage():
    """Imprime el uso actual de memoria"""
    process = psutil.Process(os.getpid())
    memory_info = process.memory_info()
    memory_gb = memory_info.rss / 1024 / 1024 / 1024
    print(f"💾 Memoria utilizada: {memory_gb:.2f} GB")

def create_temporal_features(df):
    """
    Crea características temporales a partir de la fecha (optimizado)
    """
    print("📅 Creando características temporales...")
    print_memory_usage()
    
    # Trabajar directamente sobre el DataFrame original para ahorrar memoria
    # Características básicas de tiempo
    df['year'] = df['periodo'].dt.year
    df['month'] = df['periodo'].dt.month
    df['quarter'] = df['periodo'].dt.quarter
    df['is_december'] = (df['month'] == 12).astype('int8')  # Usar int8 para ahorrar memoria
    df['is_january'] = (df['month'] == 1).astype('int8')
    
    # Tiempo transcurrido desde inicio
    min_date = df['periodo'].min()
    df['months_since_start'] = ((df['periodo'] - min_date).dt.days / 30.44).astype('int16')
    
    # Características cíclicas (usar float32 en lugar de float64)
    df['month_sin'] = (np.sin(2 * np.pi * df['month'] / 12)).astype('float32')
    df['month_cos'] = (np.cos(2 * np.pi * df['month'] / 12)).astype('float32')
    df['quarter_sin'] = (np.sin(2 * np.pi * df['quarter'] / 4)).astype('float32')
    df['quarter_cos'] = (np.cos(2 * np.pi * df['quarter'] / 4)).astype('float32')
    
    # Limpiar memoria
    gc.collect()
    
    print(f"   ✅ Características temporales creadas")
    print_memory_usage()
    return df

def create_lag_features_chunked(df, target_col='tn', lags=[1, 2, 3, 6, 12], chunk_size=100000):
    """
    Crea características de lag procesando por chunks para optimizar memoria
    """
    print(f"⏱️ Creando características de lag para {target_col} (procesamiento por chunks)...")
    print_memory_usage()
    
    df = df.sort_values(['product_id', 'customer_id', 'periodo'])
    
    # Inicializar columnas de lag con NaN
    for lag in lags:
        lag_col = f'{target_col}_lag_{lag}'
        df[lag_col] = np.nan
    
    # Procesar por grupos de product_id, customer_id
    unique_combinations = df[['product_id', 'customer_id']].drop_duplicates()
    
    total_combinations = len(unique_combinations)
    processed = 0
    
    for chunk_start in range(0, total_combinations, chunk_size):
        chunk_end = min(chunk_start + chunk_size, total_combinations)
        chunk_combinations = unique_combinations.iloc[chunk_start:chunk_end]
        
        # Crear máscara para el chunk actual
        mask = df.set_index(['product_id', 'customer_id']).index.isin(
            chunk_combinations.set_index(['product_id', 'customer_id']).index
        )
        
        chunk_df = df[mask].copy()
        
        # Aplicar lags al chunk
        for lag in lags:
            lag_col = f'{target_col}_lag_{lag}'
            chunk_df[lag_col] = chunk_df.groupby(['product_id', 'customer_id'])[target_col].shift(lag)
        
        # Actualizar el DataFrame principal
        df.loc[mask, [f'{target_col}_lag_{lag}' for lag in lags]] = chunk_df[[f'{target_col}_lag_{lag}' for lag in lags]]
        
        processed += len(chunk_combinations)
        if processed % (chunk_size * 5) == 0:
            print(f"   • Procesado: {processed}/{total_combinations} combinaciones")
            print_memory_usage()
            gc.collect()
        
        del chunk_df
        gc.collect()
    
    # Convertir a tipos más eficientes
    for lag in lags:
        lag_col = f'{target_col}_lag_{lag}'
        df[lag_col] = df[lag_col].astype('float32')
    
    print(f"   ✅ {len(lags)} características de lag creadas")
    print_memory_usage()
    return df

def create_rolling_features_optimized(df, target_col='tn', windows=[3, 6, 12]):
    """
    Crea características de ventanas móviles (optimizado para memoria)
    """
    print(f"📊 Creando características de ventanas móviles para {target_col}...")
    print_memory_usage()
    
    df = df.sort_values(['product_id', 'customer_id', 'periodo'])
    
    for window in windows:
        print(f"   • Procesando ventana {window}...")
        
        # Media móvil
        col_mean = f'{target_col}_rolling_mean_{window}'
        df[col_mean] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).mean()
        ).astype('float32')
        
        # Desviación estándar móvil
        col_std = f'{target_col}_rolling_std_{window}'
        df[col_std] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).std()
        ).astype('float32')
        
        # Máximo móvil
        col_max = f'{target_col}_rolling_max_{window}'
        df[col_max] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).max()
        ).astype('float32')
        
        gc.collect()
        print_memory_usage()
    
    print(f"   ✅ {len(windows) * 3} características de ventanas móviles creadas")
    return df

def create_aggregated_features_memory_efficient(df):
    """
    Crea características agregadas de forma eficiente en memoria
    """
    print("🎯 Creando características agregadas...")
    print_memory_usage()
    
    # Características por producto
    print("   • Agregaciones por producto...")
    product_stats = df.groupby('product_id')['tn'].agg([
        'mean', 'std', 'median', 'sum', 'count'
    ]).reset_index()
    product_stats.columns = ['product_id'] + [f'product_{col}' for col in product_stats.columns[1:]]
    
    # Convertir a tipos más eficientes
    for col in product_stats.columns[1:]:
        if col != 'product_count':
            product_stats[col] = product_stats[col].astype('float32')
        else:
            product_stats[col] = product_stats[col].astype('int32')
    
    df = df.merge(product_stats, on='product_id', how='left')
    del product_stats
    gc.collect()
    
    # Características por cliente
    print("   • Agregaciones por cliente...")
    customer_stats = df.groupby('customer_id')['tn'].agg([
        'mean', 'std', 'median', 'sum', 'count'
    ]).reset_index()
    customer_stats.columns = ['customer_id'] + [f'customer_{col}' for col in customer_stats.columns[1:]]
    
    # Convertir a tipos más eficientes
    for col in customer_stats.columns[1:]:
        if col != 'customer_count':
            customer_stats[col] = customer_stats[col].astype('float32')
        else:
            customer_stats[col] = customer_stats[col].astype('int32')
    
    df = df.merge(customer_stats, on='customer_id', how='left')
    del customer_stats
    gc.collect()
    
    # Características por combinación producto-cliente (más selectivas)
    print("   • Agregaciones por producto-cliente...")
    product_customer_stats = df.groupby(['product_id', 'customer_id'])['tn'].agg([
        'mean', 'count'  # Solo las más importantes para ahorrar memoria
    ]).reset_index()
    product_customer_stats.columns = ['product_id', 'customer_id'] + [f'prod_cust_{col}' for col in product_customer_stats.columns[2:]]
    
    # Convertir a tipos más eficientes
    product_customer_stats['prod_cust_mean'] = product_customer_stats['prod_cust_mean'].astype('float32')
    product_customer_stats['prod_cust_count'] = product_customer_stats['prod_cust_count'].astype('int32')
    
    df = df.merge(product_customer_stats, on=['product_id', 'customer_id'], how='left')
    del product_customer_stats
    gc.collect()
    
    print(f"   ✅ Características agregadas creadas")
    print_memory_usage()
    return df

def create_interaction_features_lightweight(df):
    """
    Crea características de interacción esenciales
    """
    print("🔄 Creando características de interacción...")
    print_memory_usage()
    
    # Solo las interacciones más importantes
    if 'product_mean' in df.columns:
        # Ratio de stock vs demanda histórica
        if 'stock_final' in df.columns:
            df['stock_vs_demand_ratio'] = (df['stock_final'] / (df['product_mean'] + 1e-8)).astype('float32')
        
        # Tendencia reciente vs histórica
        if 'tn_rolling_mean_3' in df.columns:
            df['recent_vs_historical'] = (df['tn_rolling_mean_3'] / (df['product_mean'] + 1e-8)).astype('float32')
    
    # Interacción mes-marca (solo si no es demasiado grande)
    if 'brand' in df.columns:
        unique_brands = df['brand'].nunique()
        if unique_brands < 1000:  # Solo si no hay demasiadas marcas
            df['brand_month'] = df['brand'].astype(str) + '_' + df['month'].astype(str)
    
    gc.collect()
    print(f"   ✅ Características de interacción creadas")
    print_memory_usage()
    return df

def handle_missing_values_efficient(df):
    """
    Maneja valores faltantes de forma eficiente en memoria
    """
    print("🔧 Manejando valores faltantes...")
    print_memory_usage()
    
    # Procesar por chunks para evitar problemas de memoria
    chunk_size = 1000000  # 1M filas por chunk
    
    # Identificar columnas con valores faltantes
    missing_cols = df.columns[df.isnull().any()].tolist()
    print(f"   • Columnas con valores faltantes: {len(missing_cols)}")
    
    for col in missing_cols:
        print(f"   • Procesando: {col}")
        
        if 'lag_' in col or 'rolling_' in col:
            # Para características de lag y rolling, rellenar con 0 o media por producto
            if 'rolling_mean' in col:
                # Calcular media por producto para esta columna
                product_means = df.groupby('product_id')[col].mean()
                df[col] = df.apply(lambda row: product_means[row['product_id']] if pd.isna(row[col]) else row[col], axis=1)
            else:
                df[col] = df[col].fillna(0)
        
        elif col == 'stock_final':
            df[col] = df[col].fillna(0)
        
        elif col in ['cat1', 'cat2', 'cat3', 'brand', 'descripcion']:
            df[col] = df[col].fillna('Unknown')
        
        elif df[col].dtype in ['float64', 'float32', 'int64', 'int32']:
            df[col] = df[col].fillna(0)
        
        gc.collect()
    
    print(f"   ✅ Valores faltantes manejados")
    print_memory_usage()
    return df

def create_features_step_by_step(df):
    """
    Ejecuta feature engineering paso a paso con checkpoints
    """
    print("\n🔧 INICIANDO FEATURE ENGINEERING OPTIMIZADO")
    print("=" * 50)
    
    original_shape = df.shape
    print(f"📊 Shape inicial: {original_shape}")
    print_memory_usage()
    
    # Paso 1: Características temporales
    print("\n🔄 PASO 1: Características temporales")
    df = create_temporal_features(df)
    
    # Paso 2: Características de lag (versión optimizada)
    print("\n🔄 PASO 2: Características de lag")
    df = create_lag_features_chunked(df, lags=[1, 2, 3, 6, 12])  # Solo los lags más importantes
    
    # Paso 3: Características de rolling (solo ventanas importantes)
    print("\n🔄 PASO 3: Características de rolling")
    df = create_rolling_features_optimized(df, windows=[3, 6, 12])  # Solo 2 ventanas importantes
    
    # Paso 4: Características agregadas
    print("\n🔄 PASO 4: Características agregadas")
    df = create_aggregated_features_memory_efficient(df)
    
    # Paso 5: Características de interacción
    print("\n🔄 PASO 5: Características de interacción")
    df = create_interaction_features_lightweight(df)
    
    # Paso 6: Manejo de valores faltantes
    print("\n🔄 PASO 6: Valores faltantes")
    df = handle_missing_values_efficient(df)
    
    final_shape = df.shape
    print(f"\n📊 Shape final: {final_shape}")
    print(f"🎯 Características agregadas: {final_shape[1] - original_shape[1]}")
    print_memory_usage()
    
    # Mostrar información de las nuevas características
    original_cols = ['periodo', 'customer_id', 'product_id', 'tn', 'plan_precios_cuidados', 'cust_request_qty', 'cust_request_tn']
    new_features = [col for col in df.columns if col not in original_cols]
    print(f"\n📋 Características disponibles para modelado: {len(new_features)}")
    
    return df, new_features

# Función alternativa más ligera
def create_essential_features_only(df):
    """
    Crea solo las características más esenciales para evitar problemas de memoria
    """
    print("\n🔧 CREANDO SOLO CARACTERÍSTICAS ESENCIALES")
    print("=" * 50)
    
    original_shape = df.shape
    print(f"📊 Shape inicial: {original_shape}")
    print_memory_usage()
    
    # Solo características temporales básicas
    df['year'] = df['periodo'].dt.year.astype('int16')
    df['month'] = df['periodo'].dt.month.astype('int8')
    df['quarter'] = df['periodo'].dt.quarter.astype('int8')
    df['is_december'] = (df['month'] == 12).astype('int8')
    
    # Solo 2 lags más importantes
    df = df.sort_values(['product_id', 'customer_id', 'periodo'])
    df['tn_lag_1'] = df.groupby(['product_id', 'customer_id'])['tn'].shift(1).astype('float32')
    df['tn_lag_12'] = df.groupby(['product_id', 'customer_id'])['tn'].shift(12).astype('float32')
    
    # Solo media móvil de 3 meses
    df['tn_rolling_mean_3'] = df.groupby(['product_id', 'customer_id'])['tn'].transform(
        lambda x: x.shift(1).rolling(window=3, min_periods=1).mean()
    ).astype('float32')
    
    # Estadísticas básicas por producto
    product_stats = df.groupby('product_id')['tn'].mean().reset_index()
    product_stats.columns = ['product_id', 'product_mean']
    product_stats['product_mean'] = product_stats['product_mean'].astype('float32')
    df = df.merge(product_stats, on='product_id', how='left')
    
    # Rellenar valores faltantes
    df['tn_lag_1'] = df['tn_lag_1'].fillna(0)
    df['tn_lag_12'] = df['tn_lag_12'].fillna(0)
    df['tn_rolling_mean_3'] = df['tn_rolling_mean_3'].fillna(0)
    
    final_shape = df.shape
    print(f"\n📊 Shape final: {final_shape}")
    print(f"🎯 Características agregadas: {final_shape[1] - original_shape[1]}")
    print_memory_usage()
    
    return df



print("🚀 FUNCIONES DE FEATURE ENGINEERING OPTIMIZADAS LISTAS")
print("💡 Opciones disponibles:")
print("   1. create_features_step_by_step(df) - Versión optimizada completa")
print("   2. create_essential_features_only(df) - Solo características esenciales")
print("   3. Ejecutar paso a paso manualmente")
```

```python
# =============================================================================
# MÓDULOS ADICIONALES DE FEATURE ENGINEERING PARA LGBM
# =============================================================================

import pandas as pd
import numpy as np
import gc
from scipy import stats
from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore')

# =============================================================================
# MÓDULO 1: CARACTERÍSTICAS DE TENDENCIAS Y MOMENTUM
# =============================================================================

def create_trend_features(df, target_col='tn', windows=[3, 6, 12]):
    """
    Crea características de tendencia y momentum
    """
    print("📈 Creando características de tendencia y momentum...")
    
    df = df.sort_values(['product_id', 'customer_id', 'periodo'])
    
    for window in windows:
        print(f"   • Procesando ventana {window}...")
        
        # Tendencia lineal (slope)
        def calculate_trend(series):
            if len(series) < 2:
                return 0
            x = np.arange(len(series))
            if series.std() == 0:
                return 0
            slope, _, _, _, _ = stats.linregress(x, series)
            return slope
        
        trend_col = f'{target_col}_trend_{window}'
        df[trend_col] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=2).apply(calculate_trend, raw=False)
        ).astype('float32')
        
        # Momentum (cambio porcentual)
        momentum_col = f'{target_col}_momentum_{window}'
        df[momentum_col] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: ((x.shift(1) - x.shift(window)) / (x.shift(window) + 1e-8)) * 100
        ).astype('float32')
        
        # Aceleración (cambio de la tendencia)
        if window > 3:
            accel_col = f'{target_col}_acceleration_{window}'
            df[accel_col] = df.groupby(['product_id', 'customer_id'])[trend_col].transform(
                lambda x: x.diff()
            ).astype('float32')
        
        # Volatilidad relativa
        volatility_col = f'{target_col}_volatility_{window}'
        df[volatility_col] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
            lambda x: (x.shift(1).rolling(window=window, min_periods=1).std() / 
                      (x.shift(1).rolling(window=window, min_periods=1).mean() + 1e-8)) * 100
        ).astype('float32')
    
    # Rellenar NaN
    trend_cols = [col for col in df.columns if 'trend_' in col or 'momentum_' in col or 
                  'acceleration_' in col or 'volatility_' in col]
    for col in trend_cols:
        df[col] = df[col].fillna(0)
    
    gc.collect()
    print(f"   ✅ {len(trend_cols)} características de tendencia creadas")
    return df

# =============================================================================
# MÓDULO 2: CARACTERÍSTICAS DE ESTACIONALIDAD AVANZADA
# =============================================================================

def create_advanced_seasonal_features(df, target_col='tn'):
    """
    Crea características de estacionalidad más sofisticadas
    """
    print("🌊 Creando características de estacionalidad avanzada...")
    
    # Características de estacionalidad por producto
    seasonal_stats = df.groupby(['product_id', 'month'])[target_col].agg(['mean', 'std']).reset_index()
    seasonal_stats.columns = ['product_id', 'month', 'seasonal_mean', 'seasonal_std']
    seasonal_stats['seasonal_mean'] = seasonal_stats['seasonal_mean'].astype('float32')
    seasonal_stats['seasonal_std'] = seasonal_stats['seasonal_std'].fillna(0).astype('float32')
    
    df = df.merge(seasonal_stats, on=['product_id', 'month'], how='left')
    
    # Índice de estacionalidad
    overall_mean = df.groupby('product_id')[target_col].mean().reset_index()
    overall_mean.columns = ['product_id', 'overall_mean']
    overall_mean['overall_mean'] = overall_mean['overall_mean'].astype('float32')
    
    df = df.merge(overall_mean, on='product_id', how='left')
    df['seasonality_index'] = (df['seasonal_mean'] / (df['overall_mean'] + 1e-8)).astype('float32')
    
    # Características de día específico del año
    df['day_of_year'] = df['periodo'].dt.dayofyear.astype('int16')
    df['week_of_year'] = df['periodo'].dt.isocalendar().week.astype('int8')
    
    # Cercanía a fechas importantes
    df['days_to_year_end'] = (365 - df['day_of_year']).astype('int16')
    df['days_from_year_start'] = df['day_of_year']
    
    # Características cíclicas avanzadas
    df['week_sin'] = np.sin(2 * np.pi * df['week_of_year'] / 52).astype('float32')
    df['week_cos'] = np.cos(2 * np.pi * df['week_of_year'] / 52).astype('float32')
    df['day_year_sin'] = np.sin(2 * np.pi * df['day_of_year'] / 365).astype('float32')
    df['day_year_cos'] = np.cos(2 * np.pi * df['day_of_year'] / 365).astype('float32')
    
    # Trimestre vs histórico
    quarterly_stats = df.groupby(['product_id', 'quarter'])[target_col].mean().reset_index()
    quarterly_stats.columns = ['product_id', 'quarter', 'quarterly_mean']
    quarterly_stats['quarterly_mean'] = quarterly_stats['quarterly_mean'].astype('float32')
    
    df = df.merge(quarterly_stats, on=['product_id', 'quarter'], how='left')
    df['quarter_vs_overall'] = (df['quarterly_mean'] / (df['overall_mean'] + 1e-8)).astype('float32')
    
    del seasonal_stats, overall_mean, quarterly_stats
    gc.collect()
    
    print("   ✅ Características de estacionalidad avanzada creadas")
    return df

# =============================================================================
# MÓDULO 3: CARACTERÍSTICAS DE COMPETENCIA Y MERCADO
# =============================================================================

def create_market_features(df, target_col='tn'):
    """
    Crea características de mercado y competencia
    """
    print("🏪 Creando características de mercado y competencia...")
    
    # Market share por categoría
    if 'cat1' in df.columns:
        category_totals = df.groupby(['cat1', 'periodo'])[target_col].sum().reset_index()
        category_totals.columns = ['cat1', 'periodo', 'category_total']
        category_totals['category_total'] = category_totals['category_total'].astype('float32')
        
        df = df.merge(category_totals, on=['cat1', 'periodo'], how='left')
        df['market_share_cat1'] = (df[target_col] / (df['category_total'] + 1e-8)).astype('float32')
        
        # Ranking dentro de la categoría
        df['rank_in_category'] = df.groupby(['cat1', 'periodo'])[target_col].rank(
            method='dense', ascending=False
        ).astype('int16')
        
        del category_totals
    
    # Concentración de demanda (Gini coefficient aproximado)
    customer_concentration = df.groupby('product_id').apply(
        lambda x: (x[target_col].sort_values(ascending=False).cumsum() / x[target_col].sum()).iloc[int(len(x)*0.8)]
        if len(x) > 0 and x[target_col].sum() > 0 else 0
    ).reset_index()
    customer_concentration.columns = ['product_id', 'demand_concentration']
    customer_concentration['demand_concentration'] = customer_concentration['demand_concentration'].astype('float32')
    
    df = df.merge(customer_concentration, on='product_id', how='left')
    
    # Número de competidores activos por categoría
    if 'cat1' in df.columns:
        competitors_count = df[df[target_col] > 0].groupby(['cat1', 'periodo'])['product_id'].nunique().reset_index()
        competitors_count.columns = ['cat1', 'periodo', 'competitors_count']
        competitors_count['competitors_count'] = competitors_count['competitors_count'].astype('int16')
        
        df = df.merge(competitors_count, on=['cat1', 'periodo'], how='left')
        df['competitors_count'] = df['competitors_count'].fillna(0)
        
        del competitors_count
    
    # Índice de diversificación del cliente
    customer_diversification = df.groupby('customer_id')['cat1'].nunique().reset_index()
    customer_diversification.columns = ['customer_id', 'customer_category_diversity']
    customer_diversification['customer_category_diversity'] = customer_diversification['customer_category_diversity'].astype('int8')
    
    df = df.merge(customer_diversification, on='customer_id', how='left')
    
    del customer_concentration, customer_diversification
    gc.collect()
    
    print("   ✅ Características de mercado creadas")
    return df

# =============================================================================
# MÓDULO 4: CARACTERÍSTICAS DE ANOMALÍAS Y OUTLIERS
# =============================================================================

def create_anomaly_features(df, target_col='tn'):
    """
    Crea características basadas en detección de anomalías
    """
    print("🚨 Creando características de anomalías...")
    
    # Z-score por producto
    product_stats = df.groupby('product_id')[target_col].agg(['mean', 'std']).reset_index()
    product_stats.columns = ['product_id', 'product_mean_anom', 'product_std_anom']
    product_stats['product_std_anom'] = product_stats['product_std_anom'].fillna(0)
    product_stats = product_stats.astype({'product_mean_anom': 'float32', 'product_std_anom': 'float32'})
    
    df = df.merge(product_stats, on='product_id', how='left')
    df['z_score_product'] = ((df[target_col] - df['product_mean_anom']) / 
                            (df['product_std_anom'] + 1e-8)).astype('float32')
    
    # Indicadores de outliers
    df['is_outlier_high'] = (df['z_score_product'] > 2).astype('int8')
    df['is_outlier_low'] = (df['z_score_product'] < -2).astype('int8')
    df['outlier_magnitude'] = np.abs(df['z_score_product']).astype('float32')
    
    # Distancia a percentiles
    percentiles = df.groupby('product_id')[target_col].quantile([0.25, 0.5, 0.75]).unstack()
    percentiles.columns = ['q25', 'q50', 'q75']
    percentiles = percentiles.reset_index()
    percentiles = percentiles.astype({'q25': 'float32', 'q50': 'float32', 'q75': 'float32'})
    
    df = df.merge(percentiles, on='product_id', how='left')
    
    df['distance_to_median'] = np.abs(df[target_col] - df['q50']).astype('float32')
    df['iqr'] = (df['q75'] - df['q25']).astype('float32')
    df['relative_position'] = ((df[target_col] - df['q25']) / (df['iqr'] + 1e-8)).astype('float32')
    
    # Rolling anomaly score
    df = df.sort_values(['product_id', 'customer_id', 'periodo'])
    df['rolling_anomaly_3'] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
        lambda x: np.abs(x - x.shift(1).rolling(window=3, min_periods=1).mean())
    ).astype('float32')
    
    # Cambios súbitos
    df['sudden_change'] = df.groupby(['product_id', 'customer_id'])[target_col].transform(
        lambda x: np.abs(x - x.shift(1)) / (x.shift(1) + 1e-8)
    ).astype('float32')
    
    df['is_sudden_increase'] = (df['sudden_change'] > 2).astype('int8')
    df['is_sudden_decrease'] = (df['sudden_change'] < -0.5).astype('int8')
    
    # Rellenar NaN
    anomaly_cols = ['z_score_product', 'distance_to_median', 'relative_position', 
                   'rolling_anomaly_3', 'sudden_change']
    for col in anomaly_cols:
        df[col] = df[col].fillna(0)
    
    del product_stats, percentiles
    gc.collect()
    
    print("   ✅ Características de anomalías creadas")
    return df

# =============================================================================
# MÓDULO 5: CARACTERÍSTICAS DE CICLO DE VIDA DEL PRODUCTO
# =============================================================================

def create_lifecycle_features(df, target_col='tn'):
    """
    Crea características del ciclo de vida del producto
    """
    print("🔄 Creando características de ciclo de vida...")
    
    # Edad del producto (desde primera venta)
    product_first_sale = df[df[target_col] > 0].groupby('product_id')['periodo'].min().reset_index()
    product_first_sale.columns = ['product_id', 'first_sale_date']
    
    df = df.merge(product_first_sale, on='product_id', how='left')
    df['product_age_months'] = ((df['periodo'] - df['first_sale_date']).dt.days / 30.44).fillna(0).astype('int16')
    
    # Fase del ciclo de vida
    df['lifecycle_phase'] = pd.cut(df['product_age_months'], 
                                  bins=[-1, 3, 12, 36, 1000], 
                                  labels=['new', 'growth', 'mature', 'decline']).astype(str)
    
    # Encode lifecycle phase
    le_lifecycle = LabelEncoder()
    df['lifecycle_phase_encoded'] = le_lifecycle.fit_transform(df['lifecycle_phase']).astype('int8')
    
    # Tendencia de ventas en diferentes fases
    df['sales_trend_last_6m'] = df.groupby('product_id')[target_col].transform(
        lambda x: x.tail(6).diff().mean() if len(x) >= 6 else 0
    ).astype('float32')
    
    # Máximo histórico y tiempo desde el máximo
    product_max = df.groupby('product_id').apply(
        lambda x: x.loc[x[target_col].idxmax(), 'periodo'] if len(x) > 0 and x[target_col].max() > 0 else x['periodo'].min()
    ).reset_index()
    product_max.columns = ['product_id', 'peak_date']
    
    df = df.merge(product_max, on='product_id', how='left')
    df['months_since_peak'] = ((df['periodo'] - df['peak_date']).dt.days / 30.44).fillna(0).astype('int16')
    
    # Ratio actual vs máximo histórico
    product_peak_value = df.groupby('product_id')[target_col].max().reset_index()
    product_peak_value.columns = ['product_id', 'peak_value']
    product_peak_value['peak_value'] = product_peak_value['peak_value'].astype('float32')
    
    df = df.merge(product_peak_value, on='product_id', how='left')
    df['current_vs_peak'] = (df[target_col] / (df['peak_value'] + 1e-8)).astype('float32')
    
    # Número de picos en la demanda
    df['peak_count'] = df.groupby('product_id')[target_col].transform(
        lambda x: len([i for i in range(1, len(x)-1) if x.iloc[i] > x.iloc[i-1] and x.iloc[i] > x.iloc[i+1]])
    ).astype('int8')
    
    del product_first_sale, product_max, product_peak_value
    gc.collect()
    
    print("   ✅ Características de ciclo de vida creadas")
    return df

# =============================================================================
# MÓDULO 6: CARACTERÍSTICAS DE CROSS-SELLING Y BASKET ANALYSIS
# =============================================================================

def create_cross_selling_features(df, target_col='tn'):
    """
    Crea características de cross-selling y análisis de canasta
    """
    print("🛒 Creando características de cross-selling...")
    
    # Productos frecuentemente comprados juntos
    active_purchases = df[df[target_col] > 0]
    
    # Tamaño de canasta por cliente-período
    basket_size = active_purchases.groupby(['customer_id', 'periodo']).size().reset_index()
    basket_size.columns = ['customer_id', 'periodo', 'basket_size']
    basket_size['basket_size'] = basket_size['basket_size'].astype('int8')
    
    df = df.merge(basket_size, on=['customer_id', 'periodo'], how='left')
    df['basket_size'] = df['basket_size'].fillna(0)
    
    # Diversidad de categorías en la canasta
    if 'cat1' in df.columns:
        category_diversity = active_purchases.groupby(['customer_id', 'periodo'])['cat1'].nunique().reset_index()
        category_diversity.columns = ['customer_id', 'periodo', 'category_diversity']
        category_diversity['category_diversity'] = category_diversity['category_diversity'].astype('int8')
        
        df = df.merge(category_diversity, on=['customer_id', 'periodo'], how='left')
        df['category_diversity'] = df['category_diversity'].fillna(0)
        
        # Ratio de diversidad
        df['diversity_ratio'] = (df['category_diversity'] / (df['basket_size'] + 1e-8)).astype('float32')
        
        del category_diversity
    
    # Productos premium en la canasta (usando precio como proxy)
    if 'cust_request_tn' in df.columns and 'cust_request_qty' in df.columns:
        df['unit_price'] = (df['cust_request_tn'] / (df['cust_request_qty'] + 1e-8)).astype('float32')
        
        # Percentil de precio del producto
        price_percentiles = df.groupby('product_id')['unit_price'].quantile(0.8).reset_index()
        price_percentiles.columns = ['product_id', 'price_p80']
        price_percentiles['price_p80'] = price_percentiles['price_p80'].astype('float32')
        
        df = df.merge(price_percentiles, on='product_id', how='left')
        df['is_premium'] = (df['unit_price'] > df['price_p80']).astype('int8')
        
        # Conteo de productos premium en canasta
        premium_count = df[df['is_premium'] == 1].groupby(['customer_id', 'periodo']).size().reset_index()
        premium_count.columns = ['customer_id', 'periodo', 'premium_count']
        premium_count['premium_count'] = premium_count['premium_count'].astype('int8')
        
        df = df.merge(premium_count, on=['customer_id', 'periodo'], how='left')
        df['premium_count'] = df['premium_count'].fillna(0)
        df['premium_ratio'] = (df['premium_count'] / (df['basket_size'] + 1e-8)).astype('float32')
        
        del price_percentiles, premium_count
    
    # Patrones de compra recurrente
    customer_product_frequency = df[df[target_col] > 0].groupby(['customer_id', 'product_id']).size().reset_index()
    customer_product_frequency.columns = ['customer_id', 'product_id', 'purchase_frequency_product']
    customer_product_frequency['purchase_frequency_product'] = customer_product_frequency['purchase_frequency_product'].astype('int8')
    
    df = df.merge(customer_product_frequency, on=['customer_id', 'product_id'], how='left')
    df['purchase_frequency_product'] = df['purchase_frequency_product'].fillna(0)
    
    # Indica si es un producto habitual para el cliente
    df['is_regular_product'] = (df['purchase_frequency_product'] >= 3).astype('int8')
    
    del basket_size, customer_product_frequency
    gc.collect()
    
    print("   ✅ Características de cross-selling creadas")
    return df

# =============================================================================
# FUNCIÓN EJECUTORA MODULAR
# =============================================================================

def execute_fe_modules(df, modules=None, target_col='tn'):
    """
    Ejecuta módulos de feature engineering de manera modular
    
    Parameters:
    -----------
    df : pandas.DataFrame
        DataFrame con los datos
    modules : list
        Lista de módulos a ejecutar. Opciones:
        - 'trends': Características de tendencia y momentum
        - 'seasonal': Estacionalidad avanzada
        - 'market': Características de mercado
        - 'anomaly': Detección de anomalías
        - 'lifecycle': Ciclo de vida del producto
        - 'crosssell': Cross-selling y basket analysis
    target_col : str
        Columna objetivo
    
    Returns:
    --------
    df : pandas.DataFrame
        DataFrame con nuevas características
    new_features : list
        Lista de nuevas características creadas
    """
    
    if modules is None:
        modules = ['trends', 'seasonal', 'market', 'anomaly', 'lifecycle', 'crosssell']
    
    print(f"\n🚀 EJECUTANDO MÓDULOS DE FEATURE ENGINEERING")
    print(f"📋 Módulos seleccionados: {modules}")
    print("=" * 60)
    
    original_cols = df.columns.tolist()
    
    # Ejecutar módulos seleccionados
    if 'trends' in modules:
        print("\n🟢 MÓDULO 1: TENDENCIAS Y MOMENTUM")
        df = create_trend_features(df, target_col)
    
    if 'seasonal' in modules:
        print("\n🟢 MÓDULO 2: ESTACIONALIDAD AVANZADA")
        df = create_advanced_seasonal_features(df, target_col)
    
    if 'market' in modules:
        print("\n🟢 MÓDULO 3: MERCADO Y COMPETENCIA")
        df = create_market_features(df, target_col)
    
    if 'anomaly' in modules:
        print("\n🟢 MÓDULO 4: ANOMALÍAS Y OUTLIERS")
        df = create_anomaly_features(df, target_col)
    
    if 'lifecycle' in modules:
        print("\n🟢 MÓDULO 5: CICLO DE VIDA")
        df = create_lifecycle_features(df, target_col)
    
    if 'crosssell' in modules:
        print("\n🟢 MÓDULO 6: CROSS-SELLING")
        df = create_cross_selling_features(df, target_col)
    
    # Identificar nuevas características
    new_features = [col for col in df.columns if col not in original_cols]
    
    print(f"\n✅ PROCESO COMPLETADO")
    print(f"📊 Características agregadas: {len(new_features)}")
    print(f"🎯 Shape final: {df.shape}")
    
    # Mostrar características por módulo
    print(f"\n📋 RESUMEN DE CARACTERÍSTICAS CREADAS:")
    for module in modules:
        module_features = []
        if module == 'trends':
            module_features = [f for f in new_features if any(x in f for x in ['trend_', 'momentum_', 'acceleration_', 'volatility_'])]
        elif module == 'seasonal':
            module_features = [f for f in new_features if any(x in f for x in ['seasonal_', 'seasonality_', 'day_of_year', 'week_', 'quarter_vs_'])]
        elif module == 'market':
            module_features = [f for f in new_features if any(x in f for x in ['market_share', 'rank_in', 'concentration', 'competitors_', 'diversity'])]
        elif module == 'anomaly':
            module_features = [f for f in new_features if any(x in f for x in ['z_score', 'outlier', 'anomaly', 'sudden_', 'distance_to'])]
        elif module == 'lifecycle':
            module_features = [f for f in new_features if any(x in f for x in ['age_', 'lifecycle_', 'peak_', 'since_peak'])]
        elif module == 'crosssell':
            module_features = [f for f in new_features if any(x in f for x in ['basket_', 'diversity', 'premium_', 'regular_'])]
        
        if module_features:
            print(f"   🔹 {module.upper()}: {len(module_features)} características")
    
    return df, new_features

# =============================================================================
# FUNCIÓN DE SELECCIÓN DE CARACTERÍSTICAS IMPORTANTES
# =============================================================================

def select_important_features(df, target_col='tn', max_features=50):
    """
    Selecciona las características más importantes usando correlación y varianza
    """
    print(f"\n🎯 SELECCIONANDO CARACTERÍSTICAS IMPORTANTES (máximo {max_features})...")
    
    # Separar características numéricas
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    feature_cols = [col for col in numeric_cols if col not in [target_col, 'customer_id', 'product_id']]
    
    # Calcular correlaciones con el target
    correlations = df[feature_cols + [target_col]].corr()[target_col].abs().sort_values(ascending=False)
    correlations = correlations.drop(target_col)
    
    # Calcular varianza
    variances = df[feature_cols].var().sort_values(ascending=False)
    
    # Combinar scores (70% correlación, 30% varianza)
    normalized_corr = (correlations - correlations.min()) / (correlations.max() - correlations.min())
    normalized_var = (variances - variances.min()) / (variances.max() - variances.min())
    
    combined_score = 0.7 * normalized_corr + 0.3 * normalized_var
    combined_score = combined_score.sort_values(ascending=False)
    
    # Seleccionar top features
    selected_features = combined_score.head(max_features).index.tolist()
    
    print(f"✅ Características seleccionadas: {len(selected_features)}")
    print(f"📊 Top 10 características:")
    for i, feature in enumerate(selected_features[:10], 1):
        print(f"   {i:2d}. {feature} (score: {combined_score[feature]:.3f})")
    
    return selected_features

print("\n🎯 MÓDULOS ADICIONALES DE FEATURE ENGINEERING LISTOS")
print("💡 Uso recomendado:")
print("   # Ejecutar todos los módulos")
print("   df_enhanced, new_features = execute_fe_modules(df)")
print("   ")
print("   # Ejecutar módulos específicos")
print("   df_enhanced, new_features = execute_fe_modules(df, modules=['trends', 'seasonal', 'anomaly'])")
print("   ")
print("   # Seleccionar características importantes")
print("   important_features = select_important_features(df_enhanced, max_features=30)")
```

```python
import pandas as pd
import numpy as np
import requests
from datetime import datetime, timedelta
from typing import Dict, List, Optional, Tuple
import warnings
warnings.filterwarnings('ignore')

class ExogenousDataFetcher:
    """Clase para obtener variables exógenas de fuentes argentinas"""
    
    def __init__(self):
        self.apis = {
            'datos_argentina': 'https://apis.datos.gob.ar/series/api/',
            'bcra_estadisticas': 'https://api.estadisticasbcra.com/',
            'fred': 'https://api.stlouisfed.org/fred/'
        }
        
    def get_inflation_data(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Obtiene datos de inflación mensual de Argentina"""
        try:
            # Serie IPC INDEC desde datos.gob.ar
            url = f"{self.apis['datos_argentina']}series"
            params = {
                'ids': '148.3_INIVELNAL_DICI_M_26',  # IPC Nivel General
                'start_date': start_date,
                'end_date': end_date,
                'format': 'json'
            }
            
            response = requests.get(url, params=params)
            if response.status_code == 200:
                data = response.json()
                df = pd.DataFrame(data['data'])
                df.columns = ['fecha', 'ipc_indice']
                df['fecha'] = pd.to_datetime(df['fecha'])
                df['ipc_indice'] = pd.to_numeric(df['ipc_indice'])
                df['inflacion_mensual'] = df['ipc_indice'].pct_change() * 100
                df['inflacion_anual'] = df['ipc_indice'].pct_change(12) * 100
                return df
            else:
                print(f"Error al obtener datos de inflación: {response.status_code}")
                return self._get_mock_inflation_data(start_date, end_date)
                
        except Exception as e:
            print(f"Error en get_inflation_data: {e}")
            return self._get_mock_inflation_data(start_date, end_date)
    
    def get_exchange_rate_data(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Obtiene tipo de cambio USD/ARS del BCRA"""
        try:
            # API BCRA Estadísticas (no oficial pero funcional)
            url = f"{self.apis['bcra_estadisticas']}usd"
            response = requests.get(url)
            
            if response.status_code == 200:
                data = response.json()
                df = pd.DataFrame(data)
                df['fecha'] = pd.to_datetime(df['d'])
                df['usd_ars'] = pd.to_numeric(df['v'])
                df = df[['fecha', 'usd_ars']]
                
                # Filtrar por fechas
                start = pd.to_datetime(start_date)
                end = pd.to_datetime(end_date)
                df = df[(df['fecha'] >= start) & (df['fecha'] <= end)]
                
                # Calcular variaciones
                df['usd_ars_var_mensual'] = df['usd_ars'].pct_change(30) * 100
                df['usd_ars_ma_7'] = df['usd_ars'].rolling(7).mean()
                df['usd_ars_ma_30'] = df['usd_ars'].rolling(30).mean()
                
                return df
            else:
                return self._get_mock_exchange_rate_data(start_date, end_date)
                
        except Exception as e:
            print(f"Error en get_exchange_rate_data: {e}")
            return self._get_mock_exchange_rate_data(start_date, end_date)
    
    def get_economic_calendar(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Obtiene fechas especiales y feriados argentinos"""
        feriados_argentina = {
            '2017': ['2017-01-01', '2017-02-27', '2017-02-28', '2017-03-24', '2017-04-14', 
                    '2017-05-01', '2017-05-25', '2017-06-17', '2017-06-20', '2017-07-09',
                    '2017-08-17', '2017-10-16', '2017-11-20', '2017-12-08', '2017-12-25'],
            '2018': ['2018-01-01', '2018-02-12', '2018-02-13', '2018-03-24', '2018-03-30',
                    '2018-04-02', '2018-05-01', '2018-05-25', '2018-06-20', '2018-07-09',
                    '2018-08-17', '2018-10-15', '2018-11-19', '2018-12-08', '2018-12-24', '2018-12-25', '2018-12-31'],
            # Agregar más años según necesidad
        }
        
        eventos_especiales = {
            'dia_madre': ['2017-10-15', '2018-10-21'],  # 3er domingo octubre
            'dia_padre': ['2017-06-18', '2018-06-17'],  # 3er domingo junio
            'dia_nino': ['2017-08-20', '2018-08-19'],   # 2do domingo agosto
            'black_friday': ['2017-11-24', '2018-11-23'],
            'cyber_monday': ['2017-11-27', '2018-11-26'],
            'navidad': ['2017-12-25', '2018-12-25'],
            'ano_nuevo': ['2017-01-01', '2018-01-01']
        }
        
        # Crear DataFrame con todas las fechas
        date_range = pd.date_range(start=start_date, end=end_date, freq='D')
        df = pd.DataFrame({'fecha': date_range})
        
        # Marcar feriados
        all_feriados = []
        for year_feriados in feriados_argentina.values():
            all_feriados.extend(year_feriados)
        
        df['es_feriado'] = df['fecha'].dt.strftime('%Y-%m-%d').isin(all_feriados).astype(int)
        
        # Marcar eventos especiales
        for evento, fechas in eventos_especiales.items():
            df[f'es_{evento}'] = df['fecha'].dt.strftime('%Y-%m-%d').isin(fechas).astype(int)
        
        # Variables temporales adicionales
        df['es_fin_semana'] = df['fecha'].dt.weekday.isin([5, 6]).astype(int)
        df['dia_mes'] = df['fecha'].dt.day
        df['es_fin_mes'] = (df['fecha'].dt.day >= 28).astype(int)
        df['es_inicio_mes'] = (df['fecha'].dt.day <= 3).astype(int)
        
        return df
    
    def _get_mock_inflation_data(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Datos mock de inflación para testing"""
        date_range = pd.date_range(start=start_date, end=end_date, freq='MS')
        np.random.seed(42)
        
        df = pd.DataFrame({
            'fecha': date_range,
            'ipc_indice': np.cumsum(np.random.normal(2.5, 1.2, len(date_range))) + 100,
        })
        df['inflacion_mensual'] = df['ipc_indice'].pct_change() * 100
        df['inflacion_anual'] = df['ipc_indice'].pct_change(12) * 100
        return df
    
    def _get_mock_exchange_rate_data(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Datos mock de tipo de cambio para testing"""
        date_range = pd.date_range(start=start_date, end=end_date, freq='D')
        np.random.seed(42)
        
        df = pd.DataFrame({
            'fecha': date_range,
            'usd_ars': np.cumsum(np.random.normal(0.1, 0.5, len(date_range))) + 20,
        })
        df['usd_ars_var_mensual'] = df['usd_ars'].pct_change(30) * 100
        df['usd_ars_ma_7'] = df['usd_ars'].rolling(7).mean()
        df['usd_ars_ma_30'] = df['usd_ars'].rolling(30).mean()
        return df


class ExogenousVariablesProcessor:
    """Procesa y agrega variables exógenas al dataset principal"""
    
    def __init__(self, base_df: pd.DataFrame):
        self.base_df = base_df.copy()
        self.fetcher = ExogenousDataFetcher()
        
    def add_all_exogenous_variables(self) -> pd.DataFrame:
        """Agrega todas las variables exógenas al dataset base"""
        print("🔄 Agregando variables exógenas...")
        
        # Determinar rango de fechas
        start_date = self.base_df['periodo'].min().strftime('%Y-%m-%d')
        end_date = self.base_df['periodo'].max().strftime('%Y-%m-%d')
        
        print(f"📅 Rango de fechas: {start_date} a {end_date}")
        
        # 1. Variables económicas
        df_with_economic = self._add_economic_variables(start_date, end_date)
        
        # 2. Variables de calendario
        df_with_calendar = self._add_calendar_variables(df_with_economic, start_date, end_date)
        
        # 3. Variables derivadas
        df_final = self._add_derived_variables(df_with_calendar)
        
        print("✅ Variables exógenas agregadas exitosamente")
        return df_final
    
    def _add_economic_variables(self, start_date: str, end_date: str) -> pd.DataFrame:
        """Agrega variables económicas macro"""
        print("💰 Agregando variables económicas...")
        
        df = self.base_df.copy()
        
        # Inflación
        try:
            inflation_df = self.fetcher.get_inflation_data(start_date, end_date)
            inflation_df['periodo'] = inflation_df['fecha'].dt.to_period('M').dt.start_time
            
            df = df.merge(
                inflation_df[['periodo', 'ipc_indice', 'inflacion_mensual', 'inflacion_anual']], 
                on='periodo', 
                how='left'
            )
            print("  ✓ Datos de inflación agregados")
        except Exception as e:
            print(f"  ⚠️ Error con inflación: {e}")
        
        # Tipo de cambio (convertir a mensual)
        try:
            exchange_df = self.fetcher.get_exchange_rate_data(start_date, end_date)
            exchange_monthly = exchange_df.groupby(exchange_df['fecha'].dt.to_period('M')).agg({
                'usd_ars': 'mean',
                'usd_ars_var_mensual': 'last',
                'usd_ars_ma_7': 'mean',
                'usd_ars_ma_30': 'mean'
            }).reset_index()
            exchange_monthly['periodo'] = exchange_monthly['fecha'].dt.start_time
            
            df = df.merge(
                exchange_monthly.drop('fecha', axis=1), 
                on='periodo', 
                how='left'
            )
            print("  ✓ Datos de tipo de cambio agregados")
        except Exception as e:
            print(f"  ⚠️ Error con tipo de cambio: {e}")
        
        return df
    
    def _add_calendar_variables(self, df: pd.DataFrame, start_date: str, end_date: str) -> pd.DataFrame:
        """Agrega variables de calendario y eventos"""
        print("📅 Agregando variables de calendario...")
        
        try:
            calendar_df = self.fetcher.get_economic_calendar(start_date, end_date)
            calendar_monthly = calendar_df.groupby(calendar_df['fecha'].dt.to_period('M')).agg({
                'es_feriado': 'sum',
                'es_dia_madre': 'max',
                'es_dia_padre': 'max',
                'es_dia_nino': 'max',
                'es_black_friday': 'max',
                'es_cyber_monday': 'max',
                'es_navidad': 'max',
                'es_ano_nuevo': 'max',
                'es_fin_semana': 'sum',
                'es_fin_mes': 'sum',
                'es_inicio_mes': 'sum'
            }).reset_index()
            calendar_monthly['periodo'] = calendar_monthly['fecha'].dt.start_time
            
            df = df.merge(
                calendar_monthly.drop('fecha', axis=1), 
                on='periodo', 
                how='left'
            )
            print("  ✓ Variables de calendario agregadas")
        except Exception as e:
            print(f"  ⚠️ Error con calendario: {e}")
        
        return df
    
    def _add_derived_variables(self, df: pd.DataFrame) -> pd.DataFrame:
        """Agrega variables derivadas y transformaciones"""
        print("🔧 Creando variables derivadas...")
        
        # Variables de interacción económica
        if 'inflacion_mensual' in df.columns and 'usd_ars_var_mensual' in df.columns:
            df['presion_economica'] = df['inflacion_mensual'] + df['usd_ars_var_mensual']
            df['estabilidad_economica'] = 1 / (1 + abs(df['presion_economica']))
        
        # Índice de estacionalidad comercial
        eventos_comerciales = ['es_dia_madre', 'es_dia_padre', 'es_dia_nino', 
                              'es_black_friday', 'es_cyber_monday', 'es_navidad']
        if all(col in df.columns for col in eventos_comerciales):
            df['indice_comercial'] = df[eventos_comerciales].sum(axis=1)
        
        # Variables temporales avanzadas
        df['trimestre_comercial'] = df['quarter'].map({1: 'Q1_bajo', 2: 'Q2_medio', 
                                                      3: 'Q3_alto', 4: 'Q4_muy_alto'})
        
        # Lags de variables económicas
        for col in ['inflacion_mensual', 'usd_ars_var_mensual']:
            if col in df.columns:
                df[f'{col}_lag_1'] = df.groupby(['customer_id', 'product_id'])[col].shift(1)
                df[f'{col}_lag_3'] = df.groupby(['customer_id', 'product_id'])[col].shift(3)
        
        print("  ✓ Variables derivadas creadas")
        return df


def apply_exogenous_pipeline(df: pd.DataFrame) -> pd.DataFrame:
    """
    Función principal para aplicar el pipeline completo de variables exógenas
    
    Args:
        df: DataFrame base con estructura similar al ejemplo
        
    Returns:
        DataFrame enriquecido con variables exógenas
    """
    print("🚀 Iniciando pipeline de variables exógenas para Argentina")
    
    # Validar estructura básica
    required_cols = ['periodo', 'customer_id', 'product_id']
    if not all(col in df.columns for col in required_cols):
        raise ValueError(f"El DataFrame debe contener las columnas: {required_cols}")
    
    # Convertir periodo a datetime si no lo está
    if not pd.api.types.is_datetime64_any_dtype(df['periodo']):
        df['periodo'] = pd.to_datetime(df['periodo'])
    
    # Aplicar procesamiento
    processor = ExogenousVariablesProcessor(df)
    enriched_df = processor.add_all_exogenous_variables()
    
    # Limpiar y rellenar valores faltantes
    enriched_df = _clean_and_fill_missing(enriched_df)
    
    print("🎉 Pipeline completado exitosamente")
    print(f"📊 Nuevas columnas agregadas: {len(enriched_df.columns) - len(df.columns)}")
    
    return enriched_df


def _clean_and_fill_missing(df: pd.DataFrame) -> pd.DataFrame:
    """Limpia y rellena valores faltantes en variables exógenas"""
    print("🧹 Limpiando datos y rellenando valores faltantes...")
    
    # Forward fill para variables económicas
    economic_cols = [col for col in df.columns if any(x in col.lower() 
                    for x in ['inflacion', 'ipc', 'usd_ars', 'presion', 'estabilidad'])]
    for col in economic_cols:
        df[col] = df[col].fillna(method='ffill').fillna(method='bfill')
    
    # Rellenar variables de calendario con 0
    calendar_cols = [col for col in df.columns if col.startswith('es_') or 'indice_comercial' in col]
    for col in calendar_cols:
        df[col] = df[col].fillna(0)
    
    print("  ✓ Limpieza completada")
    return df


# Ejemplo de uso
if __name__ == "__main__":
    # Simular estructura de tu DataFrame
    sample_data = {
        'periodo': pd.date_range('2017-01-01', '2017-12-01', freq='MS'),
        'customer_id': [10001] * 12,
        'product_id': [20001] * 12,
        'tn': np.random.uniform(50, 200, 12)
    }
    
    df_sample = pd.DataFrame(sample_data)
    
    print("📋 DataFrame original:")
    print(df_sample.head())
    print(f"Columnas originales: {len(df_sample.columns)}")
    
    # Aplicar pipeline
    df_enriched = apply_exogenous_pipeline(df_sample)
    
    print("\n📋 DataFrame enriquecido:")
    print(df_enriched.head())
    print(f"Columnas finales: {len(df_enriched.columns)}")
    print(f"Nuevas variables exógenas: {list(set(df_enriched.columns) - set(df_sample.columns))}")
```

```python
# =============================================================================
# PARTE 5: MODELADO CON LIGHTGBM - VERSIÓN TEST PARA PRED NOV 2019
# =============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from lightgbm import LGBMRegressor
import lightgbm as lgb
from datetime import datetime
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split

# Asumiendo que 'data' ya está cargado
sales_with_features['periodo'] = pd.to_datetime(sales_with_features['periodo'])

# Filtrar los datos hasta octubre de 2019
data_train = sales_with_features[sales_with_features['periodo'] <= '2019-10-31'].copy()

# Definir la fecha objetivo para la predicción
target_date = pd.to_datetime('2019-12-01')

# Crear características adicionales
data_train['year'] = data_train['periodo'].dt.year
data_train['month'] = data_train['periodo'].dt.month
data_train['months_since_start'] = (data_train['periodo'] - data_train['periodo'].min()).dt.days // 30

# Definir las características a utilizar
numeric_features = [
    'sku_size', 'year', 'quarter', 'is_december', 'is_january',#'customer_id',  'month'
    'months_since_start', 'month_sin', 'month_cos', 'quarter_sin', 'quarter_cos',
    'tn_lag_3', 'tn_lag_6', 'tn_lag_12', 'tn_rolling_mean_3', 'tn_rolling_std_3',
    'tn_rolling_max_3', 'tn_rolling_mean_6', 'tn_rolling_std_6', 'tn_rolling_max_6',
    'tn_rolling_mean_12', 'tn_rolling_std_12', 'tn_rolling_max_12',
    'stock_final', 'tn_lag_1', 'tn_lag_2',
    #'product_mean', 'product_std', 'product_median', 'product_sum', 'product_count',
    #'customer_mean', 'customer_std', 'customer_median', 'customer_sum', 'customer_count',
    #'prod_cust_mean', 'prod_cust_count',
    #'stock_vs_demand_ratio', 'recent_vs_historical'
]
categorical_features = ['product_id', 'cat1', 'cat2', 'cat3', 'brand']
features = numeric_features + categorical_features

# Preparar los conjuntos de entrenamiento y validación
X = data_train[features + ['periodo']]
y = data_train['tn']

# Usar los datos hasta septiembre de 2019 para entrenamiento y octubre de 2019 para validación
train_mask = X['periodo'] < '2019-10-01'
X_train, X_val = X[train_mask], X[~train_mask]
y_train, y_val = y[train_mask], y[~train_mask]

# Eliminar 'periodo' de las características de entrenamiento
X_train = X_train.drop('periodo', axis=1)
X_val = X_val.drop('periodo', axis=1)

print("Forma de X_train:", X_train.shape)
print("Forma de X_val:", X_val.shape)

# Crear preprocesador
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# Crear pipeline
model = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', LGBMRegressor(n_estimators=10000, 
                                learning_rate=0.05, 
                                num_leaves=31, 
                                #max_bin=511,
                                random_state=42))
])

# Entrenar el modelo
model.fit(X_train, y_train)

# Función para hacer predicciones
def make_predictions(model, data, target_date, products_to_predict, features):
    last_dates = data.groupby('product_id')['periodo'].max().reset_index()
    
    pred_data = []
    for product_id in products_to_predict:
        last_date = last_dates[last_dates['product_id'] == product_id]['periodo'].iloc[0]
        product_data = data[(data['product_id'] == product_id) & (data['periodo'] == last_date)].copy()
        if not product_data.empty:
            product_data['periodo'] = target_date
            product_data['months_since_start'] += (target_date - last_date).days // 30
            product_data['year'] = target_date.year
            product_data['month'] = target_date.month
            pred_data.append(product_data)
    
    pred_data = pd.concat(pred_data, ignore_index=True)
    
    if pred_data.empty:
        print("No hay datos disponibles para la predicción")
        return pd.DataFrame(columns=['product_id', 'tn'])
    
    X_pred = pred_data[features]
    predictions = model.predict(X_pred)
    
    results = pd.DataFrame({
        'product_id': pred_data['product_id'],
        'tn': predictions
    })
    return results

# Obtener la lista de productos a predecir (todos los productos en 201910)
products_to_predict = data_train[data_train['periodo'] == '2019-10-01']['product_id'].unique()

# Hacer predicciones
predictions = make_predictions(model, data_train, target_date, products_to_predict, features)

# Agrupar predicciones por product_id
predictions_grouped = (
    predictions.groupby('product_id', as_index=False)['tn'].sum()
    .rename(columns={'tn': 'tn_predicted'})
)

# Agrupar valores reales de diciembre 2019 por product_id
real_data_201912 = (
    sales_with_features[sales_with_features['periodo'] == '2019-12-01']
    .groupby('product_id', as_index=False)['tn'].sum()
    .rename(columns={'tn': 'tn_real'})
)

# Combinar predicciones y valores reales
comparison_df = pd.merge(predictions_grouped, real_data_201912, on='product_id', how='inner')

# Calcular métricas de error
mae = (comparison_df['tn_predicted'] - comparison_df['tn_real']).abs().mean()
error = (comparison_df['tn_predicted'] - comparison_df['tn_real']).abs().sum() / comparison_df['tn_real'].sum()

print(f"Error (Suma de errores absolutos / Suma de valores reales): {error:.4f}")
print("Error Absoluto Medio:", mae)
print("Muestra de comparación:")
print(comparison_df.head())

# Graficar Predicho vs Real
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(comparison_df['tn_real'], comparison_df['tn_predicted'], alpha=0.5)
plt.plot(
    [0, max(comparison_df['tn_real'].max(), comparison_df['tn_predicted'].max())],
    [0, max(comparison_df['tn_real'].max(), comparison_df['tn_predicted'].max())],
    'r--'
)
plt.xlabel('Valores Reales')
plt.ylabel('Valores Predichos')
plt.title('Predicho vs Real - Diciembre 2019 (Agrupado por Producto)')
plt.tight_layout()
plt.show()

# Graficar histograma de errores
errors = comparison_df['tn_predicted'] - comparison_df['tn_real']
plt.figure(figsize=(10, 6))
plt.hist(errors, bins=50)
plt.xlabel('Error de Predicción')
plt.ylabel('Frecuencia')
plt.title('Distribución de Errores de Predicción')
plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# LIGHTGBM MODEL - TRAINING ON ALL DATA TO PREDICT FEBRUARY 2020
# =============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from lightgbm import LGBMRegressor
import lightgbm as lgb
from datetime import datetime
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Asumiendo que 'sales_with_features' ya está cargado
sales_with_features['periodo'] = pd.to_datetime(sales_with_features['periodo'])

# USAR TODOS LOS DATOS DISPONIBLES PARA ENTRENAMIENTO
data_train = sales_with_features.copy()

# Definir la fecha objetivo para la predicción (Febrero 2020)
target_date = pd.to_datetime('2020-02-01')

# Crear características adicionales
data_train['year'] = data_train['periodo'].dt.year
data_train['month'] = data_train['periodo'].dt.month
data_train['months_since_start'] = (data_train['periodo'] - data_train['periodo'].min()).dt.days // 30

# Definir las características a utilizar
numeric_features = [
     'sku_size', 'year', 'quarter', 'is_december', 'is_january',#'customer_id',  'month'
    'months_since_start', 'month_sin', 'month_cos', 'quarter_sin', 'quarter_cos',
    'tn_lag_3', 'tn_lag_6', 'tn_lag_12', 'tn_rolling_mean_3', 'tn_rolling_std_3',
    'tn_rolling_max_3', 'tn_rolling_mean_6', 'tn_rolling_std_6', 'tn_rolling_max_6',
    'tn_rolling_mean_12', 'tn_rolling_std_12', 'tn_rolling_max_12',
    'stock_final', 'tn_lag_1', 'tn_lag_2',
    #'product_mean', 'product_std', 'product_median', 'product_sum', 'product_count',
    #'customer_mean', 'customer_std', 'customer_median', 'customer_sum', 'customer_count',
    #'prod_cust_mean', 'prod_cust_count',
    #'stock_vs_demand_ratio', 'recent_vs_historical'
]
categorical_features = ['product_id', 'cat1', 'cat2', 'cat3', 'brand']
features = numeric_features + categorical_features

# Preparar los datos de entrenamiento (TODOS LOS DATOS)
X_train = data_train[features]
y_train = data_train['tn']

print("Forma de X_train:", X_train.shape)
print("Rango de fechas en entrenamiento:", data_train['periodo'].min(), "to", data_train['periodo'].max())

# Crear preprocesador
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# Crear pipeline
model = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', LGBMRegressor(
        n_estimators=10000, 
        learning_rate=0.05, 
        num_leaves=31, 
        random_state=42,
        #max_bin=511,  # Aumentar el límite de bins para mayor precisión
        verbose=-1  # Suprimir mensajes de LightGBM
    ))
])

print("Entrenando el modelo con todos los datos disponibles...")
# Entrenar el modelo
model.fit(X_train, y_train)
print("Modelo entrenado exitosamente!")

# Función mejorada para hacer predicciones
def make_predictions_february_2020(model, data, target_date, features):
    """
    Hace predicciones para febrero 2020 usando los últimos datos disponibles de cada producto
    """
    # Obtener la fecha más reciente para cada producto
    last_dates = data.groupby('product_id')['periodo'].max().reset_index()
    
    pred_data = []
    products_processed = 0
    
    # Obtener todos los productos únicos
    all_products = data['product_id'].unique()
    
    for product_id in all_products:
        # Obtener la última fecha disponible para este producto
        last_date = last_dates[last_dates['product_id'] == product_id]['periodo'].iloc[0]
        
        # Obtener los datos más recientes de este producto
        product_data = data[(data['product_id'] == product_id) & 
                           (data['periodo'] == last_date)].copy()
        
        if not product_data.empty:
            # Actualizar las características temporales para febrero 2020
            product_data['periodo'] = target_date
            
            # Calcular cuántos meses han pasado desde la fecha original hasta febrero 2020
            months_diff = (target_date.year - last_date.year) * 12 + (target_date.month - last_date.month)
            product_data['months_since_start'] += months_diff
            product_data['year'] = target_date.year
            product_data['month'] = target_date.month
            
            pred_data.append(product_data)
            products_processed += 1
    
    print(f"Productos procesados para predicción: {products_processed}")
    
    if not pred_data:
        print("No hay datos disponibles para la predicción")
        return pd.DataFrame(columns=['product_id', 'tn_predicted'])
    
    pred_data = pd.concat(pred_data, ignore_index=True)
    
    # Hacer predicciones
    X_pred = pred_data[features]
    predictions = model.predict(X_pred)
    
    # Crear DataFrame con resultados
    results = pd.DataFrame({
        'product_id': pred_data['product_id'],
        'tn_predicted': predictions
    })
    
    return results

# Hacer predicciones para febrero 2020
print("Generando predicciones para febrero 2020...")
predictions = make_predictions_february_2020(model, data_train, target_date, features)

# Agrupar predicciones por product_id (sumar si hay múltiples registros por producto)
kaggle_submission = (
    predictions.groupby('product_id', as_index=False)['tn_predicted'].sum()
    .rename(columns={'tn_predicted': 'tn'})
)

# Asegurar que no hay valores negativos (opcional, dependiendo de tu caso)
kaggle_submission['tn'] = kaggle_submission['tn'].clip(lower=0)

print(f"Número de productos en la predicción: {len(kaggle_submission)}")
print(f"Suma total de predicciones: {kaggle_submission['tn'].sum():.2f}")
print("\nPrimeras 10 predicciones:")
print(kaggle_submission.head(10))

print("\nEstadísticas de las predicciones:")
print(kaggle_submission['tn'].describe())

# Guardar para envío a Kaggle
kaggle_submission.to_csv('submission_february_2020.csv', index=False)
print("\nArchivo 'submission_february_2020.csv' guardado exitosamente!")

# Mostrar información adicional
print(f"\nRango de predicciones: {kaggle_submission['tn'].min():.2f} - {kaggle_submission['tn'].max():.2f}")
print(f"Mediana de predicciones: {kaggle_submission['tn'].median():.2f}")

# Opcional: Visualizar la distribución de predicciones
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(kaggle_submission['tn'], bins=50, edgecolor='black')
plt.xlabel('Predicciones TN')
plt.ylabel('Frecuencia')
plt.title('Distribución de Predicciones - Febrero 2020')

plt.subplot(1, 2, 2)
plt.boxplot(kaggle_submission['tn'])
plt.ylabel('Predicciones TN')
plt.title('Box Plot de Predicciones')

plt.tight_layout()
plt.show()

# Mostrar el DataFrame final listo para Kaggle
print("\n" + "="*50)
print("DATAFRAME LISTO PARA KAGGLE SUBMISSION:")
print("="*50)
print(kaggle_submission.head(20))
```

```python
#PRIMERO ENTRNEO LA REG LINEAL 
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# Cargar datos
print("Cargando datos...")
data = pd.read_csv("../datasets/sell-in.txt", sep="\t")
products_to_predict = pd.read_csv('../datasets/product_id_apredecir201912.txt')

# Preparar dataset
print("Preparando dataset...")
df = data.groupby(['product_id', 'periodo'])['tn'].sum().reset_index()

# Crear features (lags) y target (t+2)
print("Creando features...")
df['target'] = df.groupby('product_id')['tn'].shift(-2)  # t+2 (Feb 2020)
for i in range(1, 12):
    df[f'tn_{i}'] = df.groupby('product_id')['tn'].shift(i)

# Productos mágicos para entrenar
magicos = [20002, 20003, 20006, 20010, 20011, 20018, 20019, 20021, 20026, 20028, 
           20035, 20039, 20042, 20044, 20045, 20046, 20049, 20051, 20052, 20053, 
           20055, 20008, 20001, 20017, 20086, 20180, 20193, 20320, 20532, 20612, 
           20637, 20807, 20838]

# Entrenar modelo con datos de diciembre 2018
print("Entrenando modelo...")
train_data = df[(df['periodo'] == 201812) & (df['product_id'].isin(magicos))].dropna()
features = ['tn'] + [f'tn_{i}' for i in range(1, 12)]
X_train = train_data[features]
y_train = train_data['target']

model = LinearRegression()
model.fit(X_train, y_train)

print(f"Modelo entrenado con {len(train_data)} registros")
print(f"Coeficientes: {model.coef_[:3]}...")  # Primeros 3 coeficientes

# Generar predicciones para febrero 2020
print("Generando predicciones...")
predicciones = []

# Datos de diciembre 2019 para hacer predicciones
dec_2019 = df[df['periodo'] == 201912].copy()

# Calcular promedio histórico (últimos 12 meses excluyendo ceros)
hist_data = df[df['periodo'].between(201812, 201911)]
promedios = hist_data[hist_data['tn'] > 0].groupby('product_id')['tn'].mean()

# Para cada producto a predecir
for product_id in products_to_predict['product_id']:
    
    # Si está en productos mágicos Y tiene datos completos, usar modelo
    if product_id in magicos:
        product_data = dec_2019[dec_2019['product_id'] == product_id]
        
        if not product_data.empty and not product_data[features].isnull().any().any():
            # Usar modelo de regresión
            X_pred = product_data[features]
            prediccion = model.predict(X_pred)[0]
            predicciones.append({'product_id': product_id, 'tn': max(0, prediccion)})
            continue
    
    # Si no, usar promedio histórico
    if product_id in promedios:
        prediccion = promedios[product_id]
    else:
        prediccion = 0  # Si no hay datos históricos
    
    predicciones.append({'product_id': product_id, 'tn': prediccion})

# Crear DataFrame final
submission = pd.DataFrame(predicciones)

# Verificar que tenemos todos los productos
print(f"Productos a predecir: {len(products_to_predict)}")
print(f"Predicciones generadas: {len(submission)}")
print(f"Productos con predicción > 0: {(submission['tn'] > 0).sum()}")

# Guardar submission
submission.to_csv('submission_kaggle.csv', index=False)
print("✅ Archivo 'submission_kaggle.csv' guardado exitosamente")

# Mostrar estadísticas
print(f"\nEstadísticas finales:")
print(f"- Predicciones con modelo: {len([p for p in predicciones if products_to_predict.loc[products_to_predict['product_id'] == p['product_id']].index[0] < len(magicos) if p['product_id'] in magicos])}")
print(f"- Predicciones con promedio: {len(submission) - len([p for p in predicciones if p['product_id'] in magicos])}")
print(f"- Valor promedio predicho: {submission['tn'].mean():.4f}")
print(f"- Valor máximo predicho: {submission['tn'].max():.4f}")

submission.head(10)
```
